#Web-based Data Visualization

```python
import pandas as pd
import numpy as np
# Must mount in drive or attach file to session
df = pd.read_csv("/content/sample_data/movies.csv")

print(df.head())
print(df.info())
```

```python
# Plot y=feature 10 over x=feature 6
# Plot y=feature 8 over x=feature 6
import matplotlib.pyplot as plt
good = df[df["IsGoodRating"] == 1]
bad  = df[df["IsGoodRating"] == 0]

plt.scatter(good["imdbRating"], good["WinsNoms"], color='blue', label='Good Rating (1)', s=30, marker='+')
plt.scatter(bad["imdbRating"], bad["WinsNoms"], color='red', label='Bad Rating (0)', s=30)

plt.title("WinsNoms vs imdbRating")
plt.xlabel("imdbRating")
plt.ylabel("WinsNoms")
plt.legend()
plt.show()
```

```python
good = df[df["IsGoodRating"] == 1]
bad  = df[df["IsGoodRating"] == 0]

plt.scatter(good["imdbRating"], good["Budget"], color='blue', label='Good Rating (1)', s=30, marker='+')
plt.scatter(bad["imdbRating"], bad["Budget"], color='red', label='Bad Rating (0)', s=30)

plt.title("Budget vs IMBD")
plt.xlabel("imdbRating")
plt.ylabel("Budget")
plt.legend()
plt.show()
```

## Question 2
- WinsNom vs imbdRating is the feature combination that is better at splitting the data. This is deducible from the plots as the overlap area of 'good rating' and 'bad rating' results is larger in the WinsNom vs Budget example. Implying that Winsnom vs imbdRating is better at separating the data.

```python
# Plot y=feature 10 over x=feature 6
# Plot y=feature 8 over x=feature 6
import matplotlib.pyplot as plt
good = df[df["IsGoodRating"] == 1]
bad  = df[df["IsGoodRating"] == 0]
plt.scatter(good["imdbRating"], good["imdbVotes"], color='blue', label='Good Rating (1)', s=good['WinsNoms'], marker='+')
plt.scatter(bad["imdbRating"], bad["imdbVotes"], color='red', label='Bad Rating (0)', s=bad['WinsNoms'])
plt.title("Votes vs. IMDb Rating sized by Wins+Nominations")
plt.xlabel("imdbRating")
plt.ylabel("imdbVotes")
plt.legend()
plt.show()
```

```python
# Plot y=feature 10 over x=feature 6
# Plot y=feature 8 over x=feature 6
import matplotlib.pyplot as plt
good = df[df["IsGoodRating"] == 1]
bad  = df[df["IsGoodRating"] == 0]
plt.scatter(good["imdbRating"], np.sqrt(good["WinsNoms"]), color='blue', label='Good Rating (1)', s=good['WinsNoms'], marker='+')
plt.scatter(bad["imdbRating"], np.sqrt(bad["WinsNoms"]), color='red', label='Bad Rating (0)', s=bad['WinsNoms'])
plt.title("Wins+Nominations (square-root-scaled) vs. IMDb Rating")
plt.xlabel("imdbRating")
plt.ylabel("sqrt WinsNoms")
plt.legend()
plt.show()
```

```python
# Plot y=feature 10 over x=feature 6
# Plot y=feature 8 over x=feature 6
import matplotlib.pyplot as plt
good = df[df["IsGoodRating"] == 1]
bad  = df[df["IsGoodRating"] == 0]
plt.scatter(good["imdbRating"], np.log(good["WinsNoms"]), color='blue', label='Good Rating (1)', s=good['WinsNoms'], marker='+')
plt.scatter(bad["imdbRating"], np.log(bad["WinsNoms"]), color='red', label='Good Rating (1)', s=bad['WinsNoms'])
plt.title("Wins+Nominations (log-scaled) vs. IMDb Rating")
plt.xlabel("imdbRating")
plt.ylabel("Log WinsNoms")
plt.legend()
plt.show()
```
